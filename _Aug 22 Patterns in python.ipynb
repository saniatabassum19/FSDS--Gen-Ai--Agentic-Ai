# 1.Right Angle Triangle Pattern# 

```python
print('Method 1')

for i in range(5):
    for j in range(i+1):
        print('*', end =" ")
    print()
print('Method 2')

for i in range(5):
    for j in range(5):
        if i>=j:
            print('*', end = " ")
    print()
print('Method 3')
for i in range(1,6):
    print(' * '*i)
```

# 2.Inverted Right Angled Triangle

```python
print('Method 1')
for i in range(5):
    for j in range(5-i):
        print('*', end =" ")
    print()
print('Method 2')
for i in range(5,0,-1):
    print('* '*i)
```

# 3.Pyramid Pattern

```python
print('Method 1')
for i in range(5):
    for j in range(2*i+1):
        print('*', end = " ")
    print()
```
